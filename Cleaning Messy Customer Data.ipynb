```python
# Basically what i am trying to do here to fill the categories  is to create a mapping tabe, categorize each product and and based on that table ,
#if the line item has that product ,we check the mapping table and then map the category.
```

```python
# creating a category , product mapping table 

mapping_df = messy_customer_df_wo_dupes[messy_customer_df_wo_dupes['category'].notna()]
mapping_df=mapping_df[['category','product']].drop_duplicates()

# Checkign if the product is mapped to multiple categories,  if it is mapped to one category it will show 1 , if not its mapped to mulitple categories
print(mapping_df.groupby('product')['category'].nunique().sort_values(ascending=False))
mapping_df[mapping_df['product'] == 'backpack']
```

```python
## since the products are mapped to mulitple categories , to impute this column i will fill the NaN values with "Unknown" 
messy_customer_df_wo_dupes['category'] =messy_customer_df_wo_dupes['category'].fillna('unknown')
messy_customer_df_wo_dupes['category'].isna().sum()
```

```python
messy_customer_df_wo_dupes['unit_price'] = messy_customer_df_wo_dupes['unit_price'].fillna(messy_customer_df_wo_dupes.groupby('category')['unit_price'].transform('median'))
```
